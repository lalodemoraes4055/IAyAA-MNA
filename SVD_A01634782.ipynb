++++++++ Inicia la sección de agregar texto: +++++++++++

El argumento "latin-1" se utiliza al cargar el archivo "geoplaces2.csv" para garantizar que los caracteres especiales, como tildes y símbolos específicos de ciertos idiomas, se interpreten correctamente. Esto es importante porque, si un archivo contiene caracteres como "á", "ñ" o "ç" y se intenta abrir con una codificación incorrecta (como UTF-8 sin especificar un manejo de errores), pueden aparecer errores de lectura o caracteres ilegibles, como signos de interrogación o cuadros vacíos. Por ejemplo, si el archivo contiene nombres de restaurantes en español como "El Rincón del Niño" o "Café & Tapas", una codificación inadecuada podría convertirlos en "El RincXn del NiXo" o "Caf? & Tapas", afectando la calidad de los datos. "latin-1" es una codificación ampliamente usada en archivos con texto en idiomas europeos y permite cargar los datos sin necesidad de conversiones adicionales.

## **Ejercicio - 4**

```
# Ejercicio 4:

#    Define la matriz de utilidad cuyos renglones sean los nombres de los
#    restaurantes, las columnas los IDs de los usuarios y las entradas la
#    evaluación general (rating). La llamaremos "UtMx_rating".


# ************* Inlcuye aquí tu código:*****************************

UtMx_rating = df_combinado.pivot(index='name', columns='userID', values='rating')
UtMx_rating = UtMx_rating.fillna(0)


# *********** Aquí termina la sección de agregar código *************


print('Dimensión de la matriz de Utilidad:')
print('(restaurantes, usuarios) =', (UtMx_rating.shape))
UtMx_rating.head()
```

#### **Similaridades con la matriz obtenida de la Factorización SVD Truncada:**

*   ### **Aplicaremo la Factorización SVD con los 66 primeros valores singulares a la matriz de utilidad UtMx_rating.**

*   ### **El valor de 66 se seleccionó porque como verás, esta cantidad de valores singulares es suficiente para describir al menos el 95% de la varianza de la matriz de utilidad.**

*   ### **Recordemos que la factorización SVD de una matriz $A$ tiene la forma:** $A_{m\times n} = U_{m\times m}\Sigma_{m\times n}V_{n\times n}^T$

*   ### **Usaremos la función coseno como medida de similaridad**

*   ### **Buscaremos restaurantes similares al llamado  "tacos de barbacoa enfrente del Tec"**.

```
num_componentes=66  # componentes más significativas

# Obtengamos la matriz de factores latentes de los restaurantes:
SVD_rating = TruncatedSVD(n_components=num_componentes)  # inicializamos
mat_fact_lat_restaurantes = SVD_rating.fit_transform(UtMx_rating) # SVD truncada-ndarray

print('Dimensión Vectores latentes restaurantes:', mat_fact_lat_restaurantes.shape)
print('Dimensión Traspuesta Vectores latentes usuarios:', SVD_rating.components_.shape)

# Calculemos la variabilidad acumulada de las componentes utilizadas:
print('\nVariabilidad acumulada de las componentes utilizadas: %.3f' % SVD_rating.explained_variance_ratio_[0:num_componentes].sum())

# matriz de similaridad
sim_matrix = cosine_similarity(mat_fact_lat_restaurantes)

# Restaurante de referencia:
restaurante_de_referencia = "tacos de barbacoa enfrente del Tec"
nombres_rest = UtMx_rating.T.columns  # lista de los nombres de los restaurantes.
idx_rest = list(nombres_rest).index(restaurante_de_referencia) # índice del restaurante de referencia.
sim_vector = sim_matrix[idx_rest] # Vector de similaridad del restaurante de referencia contra todos.

# Buscando las similaridades positivas:
idx = (sim_vector>0)
mejores_sim_rate = list()
for i in range(len(nombres_rest[idx])):
  mejores_sim_rate.append((sim_vector[idx][i], nombres_rest[idx][i]))

print('\nTotal de similaridades positivas encontradas:', len(mejores_sim_rate))

# Las ordenamos de mayor a menor relevancia:
mejores_sim_rate_ordenadas = sorted(mejores_sim_rate, key=lambda x:x[0], reverse=True)

print('\nMejores recomendaciones con base al restaurante: \"%s\":' % restaurante_de_referencia)
mejores_sim_rate_ordenadas[1:11]   # omitimos el primero que es el mismo restaurante de referencia
```

## **Ejercicio - 5**

### **Sistema Híbrido**

### **Utilicemos ahora un sistema híbrido, el cual consistirá en conjuntar la matriz de "factores_latentes_restaurantes" (que es un ndarray) obtenida previamente, con la matriz de factores adicionales (que también será un ndarray) formada por las columnas One-Hot-Encoder de los factores 'dress_code', 'accessibility', 'price' y 'franchise' del DataFrame data_geo.**

### **Por el momento no requerimos que la matriz híbrida sea un DataFrame, con que sea un arreglo de NumPy, ndarray, será suficiente.**

### **Los renglones de la matriz híbrida siguen representando a cada restaurante, pero ahora con la información de los vectores latentes de la factorización SVD, más los vectores OneHotEncoding de los nuevos factores que se incluyeron.**



```
# Ejercicio 5:

# Define el DataFrame "df_fact_adicionales" formada por los factores 'dress_code',
# 'accessibility', 'price' y 'franchise'.
# Después generar la matriz (ndarray), "mat_fact_adicionales_ohe", que se obtiene
# al tranformar las columnas (variables categóricas nominales) de "df_fact_adicionales"
# mediante OneHotEncoding.
# Finalmente conjuntar horizontalmente la matriz "mat_fact_lat_restaurantes"
# con "mat_fact_adicionales_ohe" para obtener la "matriz_hibrida" (ndarray).
# Toma en cuenta que "matriz_hibrida" estará formada por las columnas de "mat_fact_lat_restaurantes"
# seguida por las columnas de "mat_fact_adicionales_ohe".


# ************* Inlcuye aquí tu código:*****************************
data_geo = data_geo.sort_values(by="name").reset_index(drop=True)
df_fact_adicionales = data_geo[['dress_code', 'accessibility', 'price', 'franchise']]
encoder = OneHotEncoder(handle_unknown='ignore',drop='first', sparse_output=False)
mat_fact_adicionales_ohe = encoder.fit_transform(df_fact_adicionales)
matriz_hibrida = np.hstack((mat_fact_lat_restaurantes, mat_fact_adicionales_ohe))


# *********** Aquí termina la sección de agregar código *************


print('Dimensión de la matriz híbrida:')
matriz_hibrida.shape
```

```
# Recordemos que "restaurante_de_referencia" es "tacos de barbacoa enfrente del Tec".
# Obtener el índice del restaurante
idx = data_geo[data_geo['name'] == restaurante_de_referencia].index[0]

print('Restaurante de referencia:', restaurante_de_referencia)
print('Índice del restaurante de referencia:', idx)
```

## **Ejercicio - 6:**

### **Una vez que tenemos la matriz híbrida, con dicha matriz deberás obtener las similitudes (con respecto a la función coseno) del restaurante de referencia con relación a todos los restaurantes, ordenarlos de mayor a menor con respecto a la similitud del coseno y desplegar los primeros 10 de mayor similitud (sin incluir al restaurante de referencia mismo).**

```
# Ejercicio 6

# ************* Inlcuye aquí tu código:*****************************

# Restaurante de referencia
restaurante_de_referencia = "tacos de barbacoa enfrente del Tec"

# Encontramos el índice del restaurante de referencia en la matriz híbrida
nombres_rest = data_geo['name'].values  # Lista de los nombres de los restaurantes.
idx_rest = list(nombres_rest).index(restaurante_de_referencia)  # Índice del restaurante de referencia.

# Calculamos la matriz de similitudes utilizando la función coseno
sim_matrix = cosine_similarity(matriz_hibrida)

# Extraemos el vector de similitudes del restaurante de referencia con todos los demás
sim_vector = sim_matrix[idx_rest]

# Buscamos las similitudes positivas
idx = (sim_vector > 0)
mejores_sim_rate = list()
for i in range(len(nombres_rest[idx])):
    mejores_sim_rate.append((sim_vector[idx][i], nombres_rest[idx][i]))

# Ordenamos las similitudes de mayor a menor
mejores_sim_rate_ordenadas = sorted(mejores_sim_rate, key=lambda x: x[0], reverse=True)

# Excluimos el primer valor (el mismo restaurante de referencia)
print(f'Mejores recomendaciones con base en el restaurante: "{restaurante_de_referencia}":\n')

# Imprimir cada restaurante recomendado en una nueva línea
for similitud, restaurante in mejores_sim_rate_ordenadas[1:11]:
    print(f'Restaurante: {restaurante} - Similitud: {similitud:.4f}')



# *********** Aquí termina la sección de agregar código *************
```

En esta actividad, se compararon dos enfoques para un sistema de recomendación: uno basado únicamente en la factorización SVD y otro híbrido que combina la descomposición SVD con características adicionales de los restaurantes, como el dress code, la accesibilidad, el precio y la pertenencia a una franquicia.

El sistema basado exclusivamente en SVD se enfoca en las interacciones pasadas entre los usuarios y los restaurantes, extrayendo patrones implícitos entre ellos. Este método genera recomendaciones basadas en similitudes latentes, identificando principalmente restaurantes que han sido calificados de manera similar por los usuarios. En el caso del restaurante de referencia "tacos de barbacoa enfrente del Tec", el sistema SVD puro identificó como opciones más similares restaurantes como Vips, little pizza Emilio Portes Gil y tacos abi, con similitudes superiores a 0.93.

Por otro lado, el sistema híbrido incorpora información explícita adicional mediante One-Hot Encoding de atributos como el dress code, la accesibilidad, el precio y el tipo de franquicia. Esto permite que las recomendaciones no solo se basen en la similitud implícita entre usuarios y restaurantes, sino también en atributos específicos que pueden ser relevantes para la decisión del usuario. En esta versión mejorada, los resultados del sistema híbrido reflejan una mayor diversidad en las recomendaciones, destacando tacos abi, little pizza Emilio Portes Gil y Vips, pero también incluyendo restaurantes como puesto de gorditas, Carreton de Flautas y Migadas, y Carnitas Mata Calle 16 de Septiembre, que no fueron tan prominentes en el sistema basado en SVD puro.

La principal diferencia radica en que el sistema SVD puro puede capturar patrones complejos de interacción sin necesidad de información adicional, lo cual es ventajoso cuando los usuarios han calificado una amplia variedad de restaurantes. Sin embargo, el sistema híbrido mejora la capacidad de personalización y explicabilidad de las recomendaciones al incorporar atributos explícitos. Esto significa que, en lugar de basarse únicamente en relaciones implícitas detectadas por el modelo, el sistema ahora puede ofrecer sugerencias con una justificación más clara y comprensible para el usuario. Por ejemplo, si un usuario busca opciones económicas y de fácil acceso, el modelo híbrido podrá sugerir restaurantes con estas características, mientras que el modelo basado solo en SVD podría no capturar esta preferencia de manera directa.

La integración de atributos explícitos proporciona mayor contexto al modelo, facilitando que los usuarios comprendan por qué se les recomienda un restaurante en particular. Además, esto permite adaptar mejor las sugerencias a diferentes perfiles de usuario, alineando las recomendaciones no solo con calificaciones pasadas, sino también con preferencias individuales en términos de estilo, accesibilidad y costo.

En conclusión, aunque ambos enfoques tienen su utilidad, el sistema híbrido demuestra ser más robusto y flexible. No solo mejora la personalización de las recomendaciones al considerar tanto la historia de calificaciones como las preferencias explícitas del usuario, sino que también incrementa la explicabilidad del modelo, haciendo que las recomendaciones sean más transparentes y justificables. Mientras que el modelo basado en SVD sigue siendo una opción válida cuando la información sobre atributos no está disponible, el enfoque híbrido ofrece una mejor experiencia de usuario al proporcionar recomendaciones más alineadas con sus necesidades y preferencias.
